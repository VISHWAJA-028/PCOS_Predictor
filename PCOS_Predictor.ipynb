**DROPING THE DUPLICTAES**

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
numerical_columns = df.select_dtypes(include=['int64','float64']).columns

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
print(df.head())
```

**DROPING THE COLUMN OF PATIENT FILE NO**

```python
columns_to_drop = [
    "Patient File No."
]

df.drop(columns=columns_to_drop, inplace=True, errors='ignore')
```

```python
print(df.head())
```

```python
categorical_columns = df.select_dtypes(include=['object']).columns

for col in categorical_columns:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
print(df.head())
```

```python
columns_to_convert_to_numeric = [
    'II    beta-HCG(mIU/mL)',
    'AMH(ng/mL)'
]

for col in columns_to_convert_to_numeric:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    # Fill NaN values (created by coercion) with the median of the column
    df[col].fillna(df[col].median(), inplace=True)
```

Now that the problematic columns are numeric, we can re-run the outlier detection.

**DETECTING AND REMOVING THE OUTLIERS**




```python
Q1 = df.quantile(0.25, numeric_only=True)

Q3 = df.quantile(0.75, numeric_only=True)

IQR = Q3 - Q1

outliers = ((df < (Q1 - 1.5 * IQR)) |
            (df > (Q3 + 1.5 * IQR))).sum()

print(outliers)
```

```python
df = df[
    ~((df < (Q1 - 1.5 * IQR)) |
      (df > (Q3 + 1.5 * IQR))).any(axis=1)
]
```
